## Visualising comparison

```python
df = merged_data_inner.copy()
```

```python
# Calculate pct yearly change
df["pct_change_weighted_address"] = df.groupby("wd22cd")["weighted_address_count_ward22"].pct_change() * 100
df["pct_change_population"] = df.groupby("wd22cd")["population"].pct_change() * 100

df['pct_change_diff'] = df['pct_change_weighted_address'] - df['pct_change_population']

# Mark big discrepancies (top 25% by absolute difference)
threshold = df["pct_change_diff"].abs().quantile(0.95)
df["big_discrepancy"] = df["pct_change_diff"].abs() >= threshold

# Remove first row (NaN change values)
df_plot = df.dropna()

# Create a properly scaled size column to handle the extreme range
size_values = df_plot["pct_change_diff"].abs()

# Cap extreme outliers (use 95th percentile as max)
size_cap = size_values.quantile(0.95)
size_values_capped = size_values.clip(upper=size_cap)

# Scale to a reasonable range (5 to 20)
min_size, max_size = 5, 20
size_scaled = ((size_values_capped - size_values_capped.min()) / 
               (size_values_capped.max() - size_values_capped.min()) * 
               (max_size - min_size) + min_size)

# Create interactive scatter plot
fig = px.scatter(
    df_plot,
    x="pct_change_weighted_address",
    y="pct_change_population",
    color=df_plot["big_discrepancy"].map({True: "Big discrepancy", False: "Other"}),
    size=np.log1p(df_plot["pct_change_diff"].abs()),  # Log transform to handle extreme values, as there is a big range so the sizes need to be adjusted
    hover_data=[
        "year", 
        "pct_change_weighted_address", 
        "pct_change_population", 
        "pct_change_diff",
        "weighted_address_count_ward22",  
        "population"                      
    ],
    animation_frame="year",
    title="Yearly Change: Weighted Address Count vs Population",
    labels={
        "pct_change_weighted_address": "Pct Change in Weighted Address Count",
        "pct_change_population": "Pct Change in Population",
        "pct_change_diff": "Pct Change Difference",
        "weighted_address_count_ward22": "Address Count",
        "population": "Population"
    }
)

fig.update_layout(
    xaxis=dict(title="Change in Weighted Address Count"),
    yaxis=dict(title="Change in Population"),
    legend_title="Difference Category"
)

fig.show()
```

```python
# Check what your size values look like to adjut the size parameter
print("Size values summary:")
print(df_plot["pct_change_diff"].abs().describe())
print(f"Min: {df_plot['pct_change_diff'].abs().min()}")
print(f"Max: {df_plot['pct_change_diff'].abs().max()}")
print(f"Number of zeros/very small values: {(df_plot['pct_change_diff'].abs() < 0.1).sum()}")
```

### compare inidividual wards timeseries address vs population

```python
# Start with first ward
# Get unique wards only (remove duplicates)
wards = df[["wd22cd", "wd22nm"]].drop_duplicates().reset_index(drop=True)

first_code = wards.iloc[0]["wd22cd"]
first_data = df[df["wd22cd"] == first_code]

# Prepare figure
fig = go.Figure()

# Add traces for first ward
for metric, color in zip(["weighted_address_count_ward22", "population"], ["blue", "red"]):
    fig.add_trace(go.Scatter(
        x=first_data["year"],
        y=first_data[metric],
        mode="lines+markers",
        name=metric,
        line=dict(color=color)
    ))

# Create dropdown buttons
buttons = []
for _, row in wards.iterrows():
    ward_code = row["wd22cd"]
    ward_name = row["wd22nm"]
    ward_data = df[df["wd22cd"] == ward_code]
    buttons.append(dict(
        label=ward_name,
        method="update",
        args=[
            {"x": [ward_data["year"], ward_data["year"]],
             "y": [ward_data["weighted_address_count_ward22"], ward_data["population"]]},
            {"title": f"{ward_name} — Weighted Address Count vs Population"}
        ]
    ))

# Add dropdown to layout
fig.update_layout(
    updatemenus=[dict(
        active=0,
        buttons=buttons,
        direction="down",
        showactive=True
    )],
    title=f"{wards.iloc[0]['wd22nm']} — Weighted Address Count vs Population",
    xaxis_title="Year",
    yaxis_title="Value"
)

fig.show()
```

## view the relationships with wards that have the highest diveregence between population and addresses

```python
# Sort data
df = df.sort_values(["wd22cd", "year"])

# Normalise both metrics to first year per ward so they can be compared
# This allows us to see the relative change over time for each ward
# Normalise both metrics relative to first year per ward
df["addr_norm"] = df.groupby("wd22cd")["weighted_address_count_ward22"].transform(lambda x: x / x.iloc[0] * 100)
df["pop_norm"] = df.groupby("wd22cd")["population"].transform(lambda x: x / x.iloc[0] * 100)

# Difference between the two (% points)
df["norm_diff"] = df["addr_norm"] - df["pop_norm"]

# For each ward, find max divergence over time
ward_divergence = (
    df.groupby(["wd22cd", "wd22nm"])["norm_diff"]
    .apply(lambda x: abs(x).max())
    .reset_index(name="max_abs_diff")
)

# Sort so biggest divergences are first
ward_divergence = ward_divergence.sort_values("max_abs_diff", ascending=False)

# Flag outliers (e.g., top 10%)
threshold = ward_divergence["max_abs_diff"].quantile(0.90)
ward_divergence["outlier"] = ward_divergence["max_abs_diff"] >= threshold

ward_divergence.head(10)
```

```python
# Sort by divergence (biggest first)
sorted_div = ward_divergence.sort_values("max_abs_diff", ascending=False).reset_index(drop=True)

# Normalise series
df["addr_norm"] = df.groupby("wd22cd")["weighted_address_count_ward22"].transform(lambda x: x / x.iloc[0] * 100)
df["pop_norm"] = df.groupby("wd22cd")["population"].transform(lambda x: x / x.iloc[0] * 100)

# Start with first ward
first_code = sorted_div.iloc[0]["wd22cd"]
first_name = sorted_div.iloc[0]["wd22nm"]
first_data = df[df["wd22cd"] == first_code]

fig = go.Figure()

# Add traces for first ward
fig.add_trace(go.Scatter(
    x=first_data["year"],
    y=first_data["addr_norm"],
    mode="lines+markers",
    name="Address (norm)"
))
fig.add_trace(go.Scatter(
    x=first_data["year"],
    y=first_data["pop_norm"],
    mode="lines+markers",
    name="Population (norm)"
))

# DROPDOWN BUTTONS
buttons = []
for _, row in sorted_div.iterrows():
    code = row["wd22cd"]
    name = row["wd22nm"]
    w_data = df[df["wd22cd"] == code]
    buttons.append(dict(
        label=f"{name} (Δ={row['max_abs_diff']:.1f})",  # show divergence in label
        method="update",
        args=[
            {"x": [w_data["year"], w_data["year"]],
             "y": [w_data["addr_norm"], w_data["pop_norm"]]},
            {"title": f"{name} — Normalised Address vs Population (Max divergence: {row['max_abs_diff']:.1f})"}
        ]
    ))

# --- UPDATE LAYOUT ---
fig.update_layout(
    updatemenus=[dict(
        type="dropdown",
        direction="down",
        buttons=buttons,
        showactive=True
    )],
    title=f"{first_name} — Normalised Address vs Population (Max divergence: {sorted_div.iloc[0]['max_abs_diff']:.1f})",
    xaxis_title="Year",
    yaxis_title="Index (First Year = 100)"
)

fig.show()
```

### correlation coeffiecients

```python
# Start with first ward
first_code = wards.iloc[0]["wd22cd"]
first_data = df[df["wd22cd"] == first_code]

# Function to calculate correlation
def calculate_correlation(ward_data):
    return ward_data["weighted_address_count_ward22"].corr(ward_data["population"])

# Calculate correlation for first ward
first_correlation = calculate_correlation(first_data)

# Prepare figure with secondary y-axis
fig = go.Figure()

# Add address count trace (primary y-axis)
fig.add_trace(go.Scatter(
    x=first_data["year"],
    y=first_data["weighted_address_count_ward22"],
    mode="lines+markers",
    name="Address Count",
    line=dict(color="blue"),
    yaxis="y"
))

# Add population trace (secondary y-axis)
fig.add_trace(go.Scatter(
    x=first_data["year"],
    y=first_data["population"],
    mode="lines+markers",
    name="Population",
    line=dict(color="red"),
    yaxis="y2"
))

# Create dropdown buttons with correlation in labels and titles
buttons = []
for _, row in wards.iterrows():
    ward_code = row["wd22cd"]
    ward_name = row["wd22nm"]
    ward_data = df[df["wd22cd"] == ward_code]
    
    # Calculate correlation for this ward
    correlation = calculate_correlation(ward_data)
    
    # Determine correlation strength for color coding in label
    if abs(correlation) >= 0.8:
        strength = "Strong"
    elif abs(correlation) >= 0.5:
        strength = "Moderate"
    elif abs(correlation) >= 0.3:
        strength = "Weak"
    else:
        strength = "Very Weak"
    
    buttons.append(dict(
        label=f"{ward_name} (r={correlation:.3f} - {strength})",
        method="update",
        args=[
            {"x": [ward_data["year"], ward_data["year"]],
             "y": [ward_data["weighted_address_count_ward22"], ward_data["population"]]},
            {"title": f"{ward_name} — Correlation: {correlation:.3f} ({strength})"}
        ]
    ))

# Update layout with dual y-axes and correlation in initial title
fig.update_layout(
    updatemenus=[dict(
        active=0,
        buttons=buttons,
        direction="down",
        showactive=True
    )],
    title=f"{wards.iloc[0]['wd22nm']} — Correlation: {first_correlation:.3f}",
    xaxis_title="Year",
    yaxis=dict(
        title=dict(text="Weighted Address Count", font=dict(color="blue")),
        tickfont=dict(color="blue")
    ),
    yaxis2=dict(
        title=dict(text="Population", font=dict(color="red")),
        tickfont=dict(color="red"),
        anchor="x",
        overlaying="y",
        side="right"
    ),
    legend=dict(x=0.01, y=0.99)
)

fig.show()

# Optional: Print summary statistics of correlations
print("Correlation Summary Across All Wards:")
correlations_all = []
for _, row in wards.iterrows():
    ward_data = df[df["wd22cd"] == row["wd22cd"]]
    if len(ward_data) > 1:  # Need at least 2 points for correlation
        corr = calculate_correlation(ward_data)
        correlations_all.append({
            'ward_name': row["wd22nm"],
            'ward_code': row["wd22cd"],
            'correlation': corr
        })

corr_df = pd.DataFrame(correlations_all)
print(f"Mean correlation: {corr_df['correlation'].mean():.3f}")
print(f"Median correlation: {corr_df['correlation'].median():.3f}")
print(f"Standard deviation: {corr_df['correlation'].std():.3f}")
print(f"Range: {corr_df['correlation'].min():.3f} to {corr_df['correlation'].max():.3f}")

# Show wards with weakest correlations
weak_correlations = corr_df[corr_df['correlation'].abs() < 0.3].sort_values('correlation')
if len(weak_correlations) > 0:
    print(f"\nWards with very weak correlations (|r| < 0.3):")
    for _, row in weak_correlations.iterrows():
        print(f"  {row['ward_name']}: r = {row['correlation']:.3f}")
else:
    print("\nNo wards with very weak correlations found.")
```

```python
import ipywidgets as widgets
from IPython.display import display
import plotly.graph_objects as go

# Create dropdown widget
ward_dropdown = widgets.Combobox(
    options=wards["wd22nm"].tolist(),  # list of ward names
    description='Ward:',
    ensure_option=True,
    placeholder='Type ward name...',
    continuous_update=False
)

# Function to update plot
def update_plot(ward_name):
    # Find ward code
    ward_row = wards[wards["wd22nm"] == ward_name]
    if ward_row.empty:
        return  # invalid name
    ward_code = ward_row.iloc[0]["wd22cd"]

    # Get data for ward
    ward_data = df[df["wd22cd"] == ward_code]
    correlation = ward_data["weighted_address_count_ward22"].corr(ward_data["population"])

    fig = go.Figure()

    # Address Count
    fig.add_trace(go.Scatter(
        x=ward_data["year"],
        y=ward_data["weighted_address_count_ward22"],
        mode="lines+markers",
        name="Address Count",
        line=dict(color="blue"),
        yaxis="y"
    ))

    # Population
    fig.add_trace(go.Scatter(
        x=ward_data["year"],
        y=ward_data["population"],
        mode="lines+markers",
        name="Population",
        line=dict(color="red"),
        yaxis="y2"
    ))

    fig.update_layout(
        title=f"{ward_name} — Correlation: {correlation:.3f}",
        xaxis_title="Year",
        yaxis=dict(
            title=dict(text="Weighted Address Count", font=dict(color="blue")),
            tickfont=dict(color="blue")
        ),
        yaxis2=dict(
            title=dict(text="Population", font=dict(color="red")),
            tickfont=dict(color="red"),
            anchor="x",
            overlaying="y",
            side="right"
        ),
        legend=dict(x=0.01, y=0.99)
    )
    fig.show()

# Link dropdown to update function
def on_select(change):
    if change["name"] == "value" and change["new"]:
        update_plot(change["new"])

ward_dropdown.observe(on_select)

display(ward_dropdown)
```
