```python
import pandas as pd
import numpy as np
import pyreadr
import plotly.express as px
import plotly.graph_objects as go
```

```python
#read in old unit data at lsoa level
file_path = '/Users/user1/Downloads/ldd_backseries_dwellings_LSOA11CD.rds'

# Read the file
result = pyreadr.read_r(file_path)

# Get the data frame from the returned dictionary
lsoa_unit_data_old = result[None]  # None is the key for the unnamed object in the RDS file

# Show the first few rows
lsoa_unit_data_old.head()
```

```python
lsoa_unit_data_old
```

```python
df = lsoa_unit_data_old.copy()

# Get unique LSOA codes
lsoa_list = df['gss_code_lsoa'].unique()

# Create the figure
fig = go.Figure()

# Add a default trace (first LSOA)
default_lsoa = lsoa_list[0]
df_default = df[df['gss_code_lsoa'] == default_lsoa]
fig.add_trace(go.Scatter(
    x=df_default['year'],
    y=df_default['units'],
    mode='lines+markers',
    name=default_lsoa
))

# Create dropdown buttons
dropdown_buttons = []

for lsoa in lsoa_list:
    df_temp = df[df['gss_code_lsoa'] == lsoa]
    dropdown_buttons.append(dict(
        method='update',
        label=lsoa,
        args=[
            {'x': [df_temp['year']],
             'y': [df_temp['units']],
             'type': 'scatter'},
            {'title': f'Units Over Time — {lsoa}'}
        ]
    ))

# Update layout with dropdown menu
fig.update_layout(
    title=f'Units Over Time — {default_lsoa}',
    xaxis_title='Year',
    yaxis_title='Unit Count',
    updatemenus=[dict(
        active=0,
        buttons=dropdown_buttons,
        direction='down',
        x=0.1,
        xanchor='left',
        y=1.15,
        yanchor='top',
        showactive=True
    )]
)

fig.show()
```

```python
oa_adress_data_new_filtered = oa_adress_data_new.drop(['WD24NMW'], axis=1)
oa_adress_data_new_filtered
```

```python
# Sort the data
oa_adress_data_new_filtered = oa_adress_data_new_filtered.sort_values(by=['OA21CD', 'year'])

# Compute difference
oa_adress_data_new_filtered['oa_address_added'] = oa_adress_data_new_filtered.groupby('OA21CD')['address_count_oa'].diff()

# Replace the first NaN in each group with the original count
first_counts = oa_adress_data_new_filtered.groupby('OA21CD')['address_count_oa'].transform('first')
oa_adress_data_new_filtered['oa_address_added'] = oa_adress_data_new_filtered['oa_address_added'].fillna(first_counts).astype(int)

#ward count difference
oa_adress_data_new_filtered['ward_address_added'] = oa_adress_data_new_filtered.groupby('WD24CD')['address_count_ward'].diff()
# Replace the first NaN in each group with the original count
first_counts_ward = oa_adress_data_new_filtered.groupby('WD24CD')['address_count_ward'].transform('first')
oa_adress_data_new_filtered['ward_address_added'] = oa_adress_data_new_filtered['ward_address_added'].fillna(first_counts_ward).astype(int)
```

```python
lsoa_adress_data = lsoa_adress_data.sort_values(by=['LSOA21CD', 'year'])

# Compute annual address additions
lsoa_adress_data['address_added'] = lsoa_adress_data.groupby('LSOA21CD')['address_count_oa'].diff().fillna(0).astype(int)
```

```python
# Sort the data
lsoa_adress_data = lsoa_adress_data.sort_values(by=['LSOA21CD', 'year'])

# Compute difference
lsoa_adress_data['address_added'] = lsoa_adress_data.groupby('LSOA21CD')['address_count_oa'].diff()

# Replace the first NaN in each group with the original count
first_counts = lsoa_adress_data.groupby('LSOA21CD')['address_count_oa'].transform('first')
lsoa_adress_data['address_added'] = lsoa_adress_data['address_added'].fillna(first_counts).astype(int)
```

```python
lsoa_adress_data
```

```python
# First, standardize column names and types
df1 = lsoa_adress_data.rename(columns={
    'LSOA21CD': 'lsoa',
    'year': 'year',
    'address_count_oa': 'address_count',
    'address_added_oa': 'address_added'
})

df2 = lsoa_unit_data_old.rename(columns={
    'gss_code_lsoa': 'lsoa',
    'year': 'year',
    'units': 'unit_count'
})

# Make sure year is int and not float
df2['year'] = df2['year'].astype(int)

# Merge the two DataFrames on lsoa and year
merged_df = df1.merge(df2, on=['lsoa', 'year'], how='inner')

# Add a new column to compare
merged_df['address_minus_units'] = merged_df['address_count'] - merged_df['unit_count']
merged_df['address_to_unit_ratio'] = merged_df['address_count'] / merged_df['unit_count']

# Preview result
merged_df.head()
```

```python
# Summary stats
merged_df[['address_count','address_added','unit_count', 'address_minus_units', 'address_to_unit_ratio']].describe()
```

```python
# Use the merged DataFrame from before
# merged_df contains columns: lsoa, year, address_count, unit_count

# Get unique LSOA codes
lsoa_list = merged_df['lsoa'].unique()

# Create figure
fig = go.Figure()

# Add traces for the first LSOA as default
default_lsoa = lsoa_list[0]
df_filtered = merged_df[merged_df['lsoa'] == default_lsoa]

fig.add_trace(go.Scatter(
    x=df_filtered['year'],
    y=df_filtered['address_count'],
    mode='lines+markers',
    name='Address Count'
))

fig.add_trace(go.Scatter(
    x=df_filtered['year'],
    y=df_filtered['unit_count'],
    mode='lines+markers',
    name='Unit Count'
))

# Add dropdown menu to select LSOA
dropdown_buttons = []

for lsoa in lsoa_list:
    df_temp = merged_df[merged_df['lsoa'] == lsoa]
    
    address_trace = go.Scatter(
        x=df_temp['year'],
        y=df_temp['address_count'],
        mode='lines+markers',
        name='Address Count'
    )
    
    unit_trace = go.Scatter(
        x=df_temp['year'],
        y=df_temp['unit_count'],
        mode='lines+markers',
        name='Unit Count'
    )
    
    dropdown_buttons.append(
        dict(
            method='update',
            label=lsoa,
            args=[{'x': [df_temp['year'], df_temp['year']],
                   'y': [df_temp['address_count'], df_temp['unit_count']]},
                  {'title': f"Address vs Units over Time — {lsoa}"}]
        )
    )

# Update layout
fig.update_layout(
    title=f"Address vs Units over Time — {default_lsoa}",
    xaxis_title="Year",
    yaxis_title="Count",
    updatemenus=[{
        "buttons": dropdown_buttons,
        "direction": "down",
        "showactive": True,
        "x": 0.1,
        "xanchor": "left",
        "y": 1.15,
        "yanchor": "top"
    }]
)

fig.show()
```

```python
# Calculate address_added
merged_df = merged_df.sort_values(by=['lsoa', 'year'])
# Unique LSOA codes
lsoa_list = merged_df['lsoa'].unique()

# Create figure
fig = go.Figure()

# Default LSOA
default_lsoa = lsoa_list[0]
df_filtered = merged_df[merged_df['lsoa'] == default_lsoa]

# Add all three default traces
fig.add_trace(go.Scatter(
    x=df_filtered['year'],
    y=df_filtered['address_count'],
    mode='lines+markers',
    name='Address Count'
))

fig.add_trace(go.Scatter(
    x=df_filtered['year'],
    y=df_filtered['address_added'],
    mode='lines+markers',
    name='Address Added'
))

fig.add_trace(go.Scatter(
    x=df_filtered['year'],
    y=df_filtered['unit_count'],
    mode='lines+markers',
    name='Unit Count'
))

# Dropdown buttons
dropdown_buttons = []

for lsoa in lsoa_list:
    df_temp = merged_df[merged_df['lsoa'] == lsoa]
    
    dropdown_buttons.append(
        dict(
            method='update',
            label=lsoa,
            args=[
                {
                    'x': [df_temp['year'], df_temp['year'], df_temp['year']],
                    'y': [df_temp['address_count'], df_temp['address_added'], df_temp['unit_count']]
                },
                {
                    'title': f"Address Count, Added, and Unit Count — {lsoa}"
                }
            ]
        )
    )

# Layout
fig.update_layout(
    title=f"Address Count, Added, and Unit Count — {default_lsoa}",
    xaxis_title="Year",
    yaxis_title="Count",
    updatemenus=[{
        "buttons": dropdown_buttons,
        "direction": "down",
        "showactive": True,
        "x": 0.1,
        "xanchor": "left",
        "y": 1.15,
        "yanchor": "top"
    }]
)

fig.show()
```

```python
#Compute correlation per LSOA
correlation_per_lsoa = merged_df_filtered.groupby('lsoa').apply(
    lambda g: g['address_added'].corr(g['unit_count'])
).reset_index(name='correlation')


merged_df_with_corr = merged_df_filtered.merge(correlation_per_lsoa, on='lsoa')
```

```python
merged_df_with_corr
```

```python
fig = px.histogram(merged_df_with_corr, x="correlation")
fig.show()
```

```python
# Find most and least similar LSOAs
most_similar = merged_df_with_corr.sort_values('correlation', ascending=False).head()
most_different = merged_df_with_corr.sort_values('correlation', ascending=True).head()

print("Most Similar LSOAs:")
print(most_similar)

print("\nMost Different LSOAs:")
print(most_different)
```
